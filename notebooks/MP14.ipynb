## Intuition and definitions
Backpropagation applies the chain rule. For y=w*x+b and mean squared loss, dL/dw is mean(2*(y-t)*x). No hidden estimator is required.

## Worked example
With x=2, w=1, b=0, target=3, dL/dw=-4.

Conventions: NumPy images use HWC unless stated; PyTorch uses CHW or NCHW. N=batch, C=bands, H=rows, W=columns. Floating display examples use [0,1]; scientific values retain separately declared units.

```python
from pathlib import Path
import sys
root = Path.cwd()
if not (root/'src').exists():
    root = root.parent
sys.path.insert(0, str(root/'src'))
import numpy as np
import torch
from rssr.core import degrade, resize, metrics
from rssr.models import SRModel
from rssr.foundations import *
torch.manual_seed(17)
torch.set_num_threads(2)
```

```python
x=np.array([2.]); target=np.array([3.]); w=1.; b=0.
grad=np.mean(2*(w*x+b-target)*x)
eps=1e-5; loss=lambda q:np.mean((q*x+b-target)**2)
assert abs(grad-(loss(w+eps)-loss(w-eps))/(2*eps))<1e-6
print(grad)
```
