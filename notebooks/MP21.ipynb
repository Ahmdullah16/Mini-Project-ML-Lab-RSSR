## Intuition and definitions
Residual learning predicts a correction to a baseline. This eases learning near an identity-like solution but does not guarantee fidelity.

## Worked example
If the reconstruction weights and bias are zero, output equals bicubic interpolation.

Conventions: NumPy images use HWC unless stated; PyTorch uses CHW or NCHW. N=batch, C=bands, H=rows, W=columns. Floating display examples use [0,1]; scientific values retain separately declared units.

```python
from pathlib import Path
import sys
root = Path.cwd()
if not (root/'src').exists():
    root = root.parent
sys.path.insert(0, str(root/'src'))
import numpy as np
import torch
from rssr.core import degrade, resize, metrics
from rssr.models import SRModel
from rssr.foundations import *
torch.manual_seed(17)
torch.set_num_threads(2)
```

```python
m=SRModel(kind='cnn',width=8)
torch.nn.init.zeros_(m.tail[0].weight); torch.nn.init.zeros_(m.tail[0].bias)
x=torch.rand(1,3,8,8); assert torch.allclose(m(x)[0],resize(x[0],2))
print('zero residual equals baseline')
```
