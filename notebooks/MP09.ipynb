## Intuition and definitions
MSE is the mean squared sample error; PSNR = 10 log10(R^2/MSE), with declared range R. Equal images have infinite PSNR.

## Worked example
An error of 0.5 everywhere gives MSE 0.25 and PSNR 6.0206 dB for R=1.

Conventions: NumPy images use HWC unless stated; PyTorch uses CHW or NCHW. N=batch, C=bands, H=rows, W=columns. Floating display examples use [0,1]; scientific values retain separately declared units.

Predict the printed values or shape before running.

```python
from pathlib import Path
import sys
root = Path.cwd()
if not (root/'src').exists():
    root = root.parent
sys.path.insert(0, str(root/'src'))
import numpy as np
import torch
from rssr.core import degrade, resize, metrics
from rssr.models import SRModel
from rssr.foundations import *
torch.manual_seed(17)
torch.set_num_threads(2)
```

```python
r=torch.ones(3,12,12); p=r*.5
m=metrics(p,r); print(m)
assert abs(m['psnr']-6.0206)<1e-4
```
