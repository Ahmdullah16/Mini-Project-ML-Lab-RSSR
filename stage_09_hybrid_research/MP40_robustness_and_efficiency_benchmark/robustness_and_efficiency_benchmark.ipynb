```python
from pathlib import Path
import sys
root=next(p for p in (Path.cwd(),*Path.cwd().parents) if (p/'src/rssr').is_dir())
sys.path.insert(0, str(root/'src'))
import numpy as np
import torch
from rssr.core import degrade, resize, metrics
from rssr.models import SRModel
from rssr.foundations import *
torch.manual_seed(17)
torch.set_num_threads(2)
```

```python
import time
m=SRModel(kind='cnn',width=8).eval(); x=torch.rand(1,3,16,16)
with torch.inference_mode():
    for _ in range(2): m(x)
    times=[]
    for _ in range(5):
        t=time.perf_counter(); m(x); times.append(time.perf_counter()-t)
print('CPU float32 NCHW=1,3,16,16 median seconds excluding IO',float(np.median(times)))
```
